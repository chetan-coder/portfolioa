# 3.08 Iterations — Homework Submission

The popcorn hack and homework hack from the 3.08 Iterations lesson. Both use a `for` loop with `continue` to skip safe values and `break` to stop at the first critical one.

## Popcorn Hack — Wildfire dispatch scanner

Alerts under 3 are safe, so `continue` skips them. Everything else needs action and gets added to the score. As soon as an alert is 8 or higher, `break` stops the scan, so the last alert (5) never gets checked. It ends with 2 skipped, 4 actions, a score of 22, and a critical alert found.

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
alerts = [1, 4, 2, 6, 3, 9, 5]

safe_cutoff = 3
critical_cutoff = 8

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

## Homework Hack — Server load checker

Same idea but for server load. Under 20 is safe, 20 and up needs action, and 55 or more is critical. The loop breaks at 60, so the last value (35) is never checked. That's why the report says 7 checked instead of all 8.

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: server load monitoring

server_loads = [10, 25, 15, 30, 45, 20, 60, 35]

safe_cutoff = 20
critical_cutoff = 55

checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

for value in server_loads:
    checked_count += 1

    if value < safe_cutoff:
        skipped_count += 1
        print(str(value) + ": safe, no action")
        continue

    action_count += 1
    total_score += value
    print(str(value) + ": action needed")

    if value >= critical_cutoff:
        critical_found = True
        print("Critical value found. Stop checking.")
        break

print("Final Report")
print("Checked: " + str(checked_count))
print("Skipped: " + str(skipped_count))
print("Actions: " + str(action_count))
print("Total score: " + str(total_score))
print("Critical found: " + str(critical_found))
```
