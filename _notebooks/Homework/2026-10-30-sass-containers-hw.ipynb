## OCS Containers HW

Homework for the OCS SASS Containers Grammar lesson, with the Poway Rotary theme. I ran the four examples, did the Parade Invite popcorn hack, and built the Parade Hub page for the homework. Each one sits inside an `ocs__container` with no inline styles.

### Example A: Container and Card

The simplest version: one container, one card, one description. The container is what keeps it centered at a readable width instead of stretching across the whole screen.

```python
%%html
<!-- UI_RUNNER: Containers Example A -->
<div class="ocs__container">
  <h2 class="ocs__section-title">Poway Scripps Rotary Club</h2>
  <div class="ocs__card">
    <p class="ocs__description">Neighbors serving Poway together through weekly fellowship and local service projects.</p>
  </div>
</div>
```

### Example B: Adding a Grid

Same idea but with a 2 column grid inside the card. The two cells sit side by side and stack into one column on a small screen.

```python
%%html
<!-- UI_RUNNER: Containers Example B -->
<div class="ocs__container">
  <div class="ocs__card">
    <h3 class="ocs__section-title">Rotary Service Projects</h3>
    <div class="ocs__grid ocs__grid--standard cols-2">
      <div class="ocs__grid-cell">Parade: The club hosts the annual Poway Rotary Parade on Poway Road.</div>
      <div class="ocs__grid-cell">Service: Members fund scholarships, park cleanups, and food drives.</div>
    </div>
  </div>
</div>
```

### Example C: Full Container Composition

Everything from the "How it Fits" diagram in one card: a grid with an accent cell, a table, and a callout at the bottom.

```python
%%html
<!-- UI_RUNNER: Containers Example C -->
<div class="ocs__container">
  <h2 class="ocs__section-title">Poway Rotary Parade Hub</h2>
  <div class="ocs__card">
    <div class="ocs__grid ocs__grid--standard cols-2">
      <div class="ocs__grid-cell ocs__grid-cell--accent">Parade Day: Saturday at 9:00AM on Poway Road</div>
      <div class="ocs__grid-cell">Volunteers: Greeters, setup crew, and cleanup team</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Event</th><th>Status</th></tr>
        <tr><td>Parade Floats</td><td>Confirmed</td></tr>
        <tr><td>Food Drive</td><td>Collecting donations</td></tr>
      </table>
    </div>
    <div class="ocs__callout">One Rotary container holds the whole Parade hub, with the same styling on phones and computers and no custom CSS.</div>
  </div>
</div>
```

### Live Example: Rotary Events in a Standard Grid

Three event cells with `cols-3`. Making the window narrow drops it to fewer columns, and none of that needed extra code.

```python
%%html
<!-- UI_RUNNER: Containers Live Example -->
<div class="ocs__container">
  <div class="ocs__card">
    <h3 class="ocs__section-title">Rotary Events Resize With You</h3>
    <p class="ocs__description">One Rotary container holds every event update. The standard grid splits it into columns that stretch and wrap.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Parade gathers crowds down Poway Road.</div>
      <div class="ocs__grid-cell">Service projects fund scholarships.</div>
      <div class="ocs__grid-cell">Weekly meetings connect volunteers.</div>
    </div>
  </div>
</div>
```

## Popcorn Hack: Parade Invite Rescue

Swapped `page-wrap` for one `ocs__container`, put everything in an `ocs__card`, and turned the purple `<h1>` into an `ocs__section-title`. The description line is my own. The flex row is now a `cols-3` standard grid, and Start time is the accent cell since that's the detail people need most.

```python
%%html
<!-- UI_RUNNER: Containers Popcorn -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">Floats, bands, and local clubs on Poway Road. Bring the whole family.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell">What you get: An invaluable local and cultural experience!</div>
    </div>
  </div>
</div>
```

## Homework Hack: Parade Hub Page

Rebuilt the broken Parade Hub starter with only container grammar:

- one `ocs__container` holding one `ocs__card`
- an `ocs__section-title` and a one line `ocs__description`
- a `cols-3` standard grid with a `--header` cell, an `--accent` cell for the start time, and a `--muted` cell for float line up since it matters less to people watching
- the old `border='1'` table rebuilt with `ocs__table-wrap` and `ocs__table`, with a header row and two rows
- an `ocs__callout` at the end

I skipped the optional SCSS extra so the page stays at zero custom CSS.

```python
%%html
<!-- UI_RUNNER: Containers Homework -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need for parade day on Poway Road, all in one place.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day</div>
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Parade floats line up before the start</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Entry</th><th>Status</th></tr>
        <tr><td>Parade floats</td><td>Open</td></tr>
        <tr><td>Volunteer sign-up</td><td>Open</td></tr>
      </table>
    </div>
    <div class="ocs__callout">Check back here before parade day for any changes to the route or start time.</div>
  </div>
</div>
```
