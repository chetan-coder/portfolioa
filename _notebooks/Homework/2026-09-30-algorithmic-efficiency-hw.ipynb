| # | Question | Answer | Why |

| 1 | Binary search requires | **A** — a sorted catalog with direct access to entries | It throws away half the catalog based on the middle entry. That only works if the entries are in order and you can jump straight to the middle. |

## Homework

**Adapting Example C for 12 books.** `visits` is scanning each book once, so it equals n. `pairs` is comparing every book with every book, including itself and in both orders, so it equals n × n.

**Prediction:** 12 books, 12 visits, 144 pairs. Doubled to 24: 24 books, 24 visits, 576 pairs.


```python
# CODE_RUNNER: 3.17 Homework - Advise the librarian (12 books)

n = 12
visits = n
pairs = n * n
print("Books in catalog: " + str(n))
print("Visits to scan every book: " + str(visits))
print("Ordered pairs to compare every book: " + str(pairs))
```

```python
# CODE_RUNNER: 3.17 Homework - Double the catalog (24 books)

n = 24
visits = n
pairs = n * n
print("Books in catalog: " + str(n))
print("Visits to scan every book: " + str(visits))
print("Ordered pairs to compare every book: " + str(pairs))
```

**Actual output:** `12, 12, 144`, then `24, 24, 576`. Matches the prediction.

**Why the growth is different.** Visits are **linear**: doubling the books from 12 to 24 doubles the visits from 12 to 24. Pairs are **quadratic**: doubling the books quadruples the pairs, 144 × 4 = 576, because each of the 24 books is now compared against 24 books instead of 12. That quadratic task gets expensive fast, but it can't be swapped for a scan if pairs are what the librarian actually needs.

**Recommendation for 1,000 books** (checkpoint B's calculator, below).


```python
# CODE_RUNNER: 3.17 Homework - Recommend a search for 1,000 books
import math

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print("Books in catalog: " + str(n))
print("Linear search, worst case: " + str(linear_checks) + " checks")
print("Binary search, worst case: " + str(binary_checks) + " checks")
```

**Actual output:** 1,000 books, 1,000 linear checks, 10 binary checks.

**What the empty test proves.** Visits and pairs handle 0 without any special case, since 0 and 0 × 0 are both 0. Binary search is different: `math.log2(0)` is undefined and crashes with `ValueError: math domain error`. The `0 if n == 0 else ...` guard is the only reason this test passes. The main path never touches that guard, so only the empty catalog shows whether it's there.
