## SASS Button Grammar HW

Homework for the SASS Buttons Grammar lesson, which uses the My Good Brain mental health site as the theme. Below are the lesson examples, all three popcorn hacks, and both homework hacks. There are no `style` attributes or made up classes, just `<button>` and `<a>` tags with OCS classes.

### Lesson Examples

I put the example snippets from 5.1 to 5.3 into one runner to compare them. The plain `ocs__btn` link is the most low key, `pill accent fill` stands out the most, and the `alert-*` tones read kind of like a traffic light. I swapped the empty and `/resources` links for the real My Good Brain page so they go somewhere.

```python
%%html
<!-- UI_RUNNER: Buttons Examples -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Resource Directory</a>
<button class="ocs__btn pill accent fill">Log Daily Mood</button>

<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <button class="ocs__btn outline">Save Draft Journal</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>

<div class="ocs__links--wide">
  <button class="ocs__btn alert-green fill">Complete A Mental Health Check-In Form</button>
  <a class="ocs__btn alert-yellow outline" href="https://www.mygoodbrain.org/resources">Request Break</a>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Popcorn Hack 1: Add the Base Button Class

The class was empty, so I added `ocs__btn`. That's all it takes to turn the plain link into a button.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 1 -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Mental Health Resource Directory</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

Added `pill` and `fill` to Complete Daily Check-In, then changed the plain `<div>` into an `ocs__links` container so both actions sit together with even spacing.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 -->
<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">Save Draft Journal</a>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout Bar

The task says to change the second link from `alert-yellow` to `alert-green`, but the yellow one is actually the Complete Check-In button, so that's the one I switched. Then I changed the container to `ocs__links--wide` and added `fill` to Get Crisis Support so it has full priority.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 3 -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green outline">Complete Check-In</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

Each category gets a tone that matches how urgent it is: red for Urgent Support, green for Self-Care, and an accent outline for Community. All three are pills inside one `ocs__links`. I also fixed the hrefs. The starter had markdown link syntax pasted inside them, so the links wouldn't have worked.

```python
%%html
<!-- UI_RUNNER: Buttons Homework 1 -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

Log Mood Entry is the main action, so it's the only filled button. View Weekly Trends is a plain outline, and the hotline is a red outline so it still stands out without fighting the main button. The requirements call the third action Crisis Hotline, so I renamed it from Mental Health Blog and linked it to the 988 Lifeline.

```python
%%html
<!-- UI_RUNNER: Buttons Homework 2 -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://988lifeline.org/">Crisis Hotline</a>
</div>
```
