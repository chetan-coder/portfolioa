## MCQ

**MCQ 3.1: 4/4 | answers: B,A,C,B**

| # | Question | Answer | Why |
|---|---|---|---|
| 1 | Best type for player tag `"007"` | **B — string** | An int would drop the leading zeros and print `7`. A tag is a label, not a quantity — you never do math on it. |
| 2 | After `stars ← 8`, `checkpoint ← stars`, `stars ← 12`, what is `checkpoint`? | **A — 8** | The assignment copied the value at that moment. Changing `stars` afterward doesn't reach back into `checkpoint`. |
| 3 | Which AP operator assigns a value? | **C — ←** | In College Board pseudocode `←` assigns and `=` tests equality. Python is the reverse: `=` assigns and `==` compares. |
| 4 | When does a previously stored total change? | **B — when reassigned/recalculated** | A total is a stored result, not a live formula. It holds whatever number it was given until a new assignment replaces it. |


## Homework



```python
# CODE_RUNNER: UESL 3.1 Homework - Star counter with a bonus

game_title = "Star Trail: Poway Run"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars

print(game_title)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

# Bonus of 5 on level two
bonus = 5
level_two_stars = level_two_stars + bonus

# The stored total does NOT change on its own
print("Stored total after bonus:", total_stars)

# It only changes when we recalculate it
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated total:", total_stars)
```
