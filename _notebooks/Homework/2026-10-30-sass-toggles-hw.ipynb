### Risk Check Example

Each toggle that's on adds 25% when you press Check Risk. I changed two things. The button got `ocs__btn fill` so it matches the rest of the site, and the script only looks at the toggles inside this example. The original used `document.querySelectorAll(".ocs__toggle-input")`, which would also count the toggles from my other cells on this page.

```python
%%html
<!-- UI_RUNNER: Toggle risk example | show_code: true -->
<div id="risk-check">
  <label class="ocs__toggle ocs__toggle--green">
    <input id="risk-1" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you have family members who have or have had cancer?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="risk-2" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you consume alcohol or use tobacco?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="risk-3" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you forget to wear sunscreen?</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="risk-4" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Do you consume less than 48 oz of water and 30 grams of fiber a day?</span>
  </label>

  <br>
  <button id="check-risk" class="ocs__btn fill">Check Risk</button>
  <p id="risk-output"></p>
</div>
<script>
function setUpRiskCheck() {
  document.getElementById("check-risk").addEventListener("click", function () {
    // start from 0 every time the button is clicked
    let risk = 0;
    const toggles = document.getElementById("risk-check").querySelectorAll(".ocs__toggle-input");
    toggles.forEach(function (toggle) {
      if (toggle.checked) {
        risk += 25;
      }
    });
    document.getElementById("risk-output").textContent = "Current Cancer Exposure Risk: " + risk + "%";
  });
}
setUpRiskCheck();
</script>
```

### Find the Bug

The track `<span>` had an empty class, so there was nothing to draw the actual switch. Adding `ocs__toggle-track` fixed it, and I added `aria-hidden="true"` too since the track is only there for looks. The check under it is the same one from the lesson.

```python
%%html
<!-- UI_RUNNER: Toggle find the bug -->
<label id="bug-toggle" class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
<p id="bug-result"></p>
<script>
function checkBugFix() {
  const fixed = document.getElementById("bug-toggle").querySelector(".ocs__toggle-track");
  document.getElementById("bug-result").textContent = fixed ? "Toggle works! 🎉" : "Incorrect, almost there!";
}
checkBugFix();
</script>
```

## Homework: Senior Settings Panel

Three toggles for a settings panel on the Friends of Poway Seniors site, and each one does something you can see:

- **Show this week's events** shows or hides the event list
- **Bigger text** makes the text in the panel bigger
- **Bingo reminder** changes the reminder message

The line at the top counts how many are on and updates every time one flips. My comments are in the code.

```python
%%html
<!-- UI_RUNNER: Toggles Homework -->
<div id="senior-settings" class="ocs__card">
  <h3>My Settings</h3>
  <p id="toggles-on">Toggles on: 0 of 3</p>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="show-events" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show this week's events</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="big-text" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Bigger text</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="bingo-reminder" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Bingo reminder</span>
  </label>

  <ul id="event-list" hidden>
    <li>Tuesday: Bingo and lunch, doors open at 9:00 AM</li>
    <li>Sunday: Bingo, games start at 12:30 PM</li>
  </ul>
  <p id="reminder-status">No Bingo reminder set.</p>
</div>
<script>
function setUpSeniorSettings() {
  const panel = document.getElementById("senior-settings");
  // I only grab the toggles inside this panel. If I used document.querySelectorAll
  // like the risk example, it would count the toggles from my other cells too.
  const toggles = panel.querySelectorAll(".ocs__toggle-input");
  const showEvents = document.getElementById("show-events");
  const bigText = document.getElementById("big-text");
  const bingoReminder = document.getElementById("bingo-reminder");

  function update() {
    // .checked is just true or false, so each toggle works like the IF/ELSE
    // in the lesson's pseudocode: on does one thing, off does the other
    document.getElementById("event-list").hidden = !showEvents.checked;
    panel.style.fontSize = bigText.checked ? "1.25rem" : "";
    document.getElementById("reminder-status").textContent = bingoReminder.checked
      ? "Reminder set for Bingo on Tuesday and Sunday."
      : "No Bingo reminder set.";

    // count how many are on, same idea as adding up the risk
    let count = 0;
    toggles.forEach(function (toggle) {
      if (toggle.checked) {
        count += 1;
      }
    });
    document.getElementById("toggles-on").textContent = "Toggles on: " + count + " of " + toggles.length;
  }

  // "change" fires every time a switch flips, so the page updates right away
  // and I don't need a Check button like the risk example has
  toggles.forEach(function (toggle) {
    toggle.addEventListener("change", update);
  });
  update();
}
setUpSeniorSettings();
</script>
```
