### Tech Talk: Changing the fr

`2fr 1fr` kept two boxes but made Beaker 1 twice as wide. `1fr 3fr` gave Beaker 2 three of the four shares. For the last step I switched to `1fr 1fr 1fr` and added a third beaker, so now there are three equal boxes. The number of values is how many boxes you get, and the size of each number is how wide that box is.

```python
%%html
<!-- UI_RUNNER: Grids Tech Talk fr -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
</div>
```

## Homework Hack

Grid 1 is a standard grid. The title is a `--header` cell and Beaker 3 is the `--accent` since 31.6 is way hotter than the other two. Grid 2 uses the `--calculator` variant with `Avg` as an accent key. No styles or made up classes are left.

I kept the lesson's PASS/FIX checklist at the top. The only difference is it reads the classes from my grids on the page instead of from a string.

```python
%%html
<!-- UI_RUNNER: Grids Homework -->
<ul id="grids-hw-checks"></ul>
<div id="grids-hw">
  <div class="ocs__grid ocs__grid--standard">
    <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
    <div class="ocs__grid-cell">Beaker 1: 22.4</div>
    <div class="ocs__grid-cell">Beaker 2: 22.9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
  </div>
  <div class="ocs__grid ocs__grid--calculator">
    <div class="ocs__grid-cell">7</div>
    <div class="ocs__grid-cell">8</div>
    <div class="ocs__grid-cell">9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
  </div>
</div>
<script>
function checkGridsHomework() {
  // Same checks as the lesson's homework runner
  let myHTML = document.getElementById("grids-hw").innerHTML;
  let classList = [];
  for (let found of (myHTML.match(/class="[^"]*"/g) || [])) {
    classList = classList.concat(found.slice(7, -1).split(" ").filter(c => c !== ""));
  }
  let madeUp = classList.filter(c => !c.startsWith("ocs__") && !c.startsWith("cols-"));

  function line(label, ok) { return "<li>" + (ok ? "PASS  " : "FIX  ") + label + "</li>"; }
  document.getElementById("grids-hw-checks").innerHTML =
    line("No inline style attributes", !myHTML.includes("style=")) +
    line("No made up classes" + (madeUp.length > 0 ? " (found: " + madeUp.join(", ") + ")" : ""), madeUp.length === 0) +
    line("Has an ocs__grid root", classList.includes("ocs__grid")) +
    line("Uses the standard variant for Grid 1", classList.includes("ocs__grid--standard")) +
    line("Uses the calculator variant for Grid 2", classList.includes("ocs__grid--calculator")) +
    line("Every box uses ocs__grid-cell", classList.filter(c => c === "ocs__grid-cell").length >= 8) +
    line("Title is a header cell", classList.includes("ocs__grid-cell--header")) +
    line("Has accent cells (outlier beaker and Avg key)", classList.filter(c => c === "ocs__grid-cell--accent").length >= 2);
}
checkGridsHomework();
</script>
```
