## Popcorn Hack 1 — Simple Procedure Call & Parameters

`report_status` is defined once with two parameters, `location` and `status`. It is then called twice with different arguments, so the same instructions print two different updates.


```python
# CODE_RUNNER: 3.12 Popcorn 1 - report_status with parameters

def report_status(location, status):
    print("Status update from " + location + ": " + status)


report_status("Poway", "Road cleared after fallen tree")
report_status("Rancho Bernardo", "Shelter open at the community center")
```

## Popcorn Hack 2 — Decomposition & Helper Procedures

`is_priority` is the helper. It does one sub-task, deciding whether an incident is a priority, and **returns** a Boolean instead of printing.

`process_report` is the outer procedure. It calls the helper, stores the returned value in `priority`, and uses an if/else to pick the message. The outer procedure never has to know *how* priority is decided. That is the decomposition.


```python
# CODE_RUNNER: 3.12 Popcorn 2 - helper procedure inside an outer procedure

# Helper Procedure
def is_priority(incident_type):
    return incident_type == "Medical Emergency"


# Outer Procedure
def process_report(location, incident_type):
    priority = is_priority(incident_type)

    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)


process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack — Emergency Dispatch Program




```python
# CODE_RUNNER: 3.12 Homework - emergency dispatch with a helper procedure

# Helper Procedure: decides priority, returns True or False
def check_priority(incident_type):
    return (incident_type == "Medical Emergency"
            or incident_type == "Fire"
            or incident_type == "Scam Call to Senior")


# Outer Procedure: calls the helper, then decides what to print
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)

    if priority:
        print("DISPATCH NOW: " + incident_type + " in " + location + " - send the nearest team.")
    else:
        print("LOGGED: " + incident_type + " in " + location + " - handled in normal order.")


dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Poway", "Scam Call to Senior")
dispatch_response("Scripps Ranch", "Fallen Tree")
```
